```python
import axelrod as axl
from agents.aif.jax.five_state import JaxFiveStateAgent
from agents.aif.jax.five_state_noise import JaxFiveStateAgentNoisy
from agents.aif.jax.factorized import JaxFactorizedAgent
import os
```

```python
# DBS vs Heuristics

DBS_vs = [
    axl.TitForTat(),
    axl.WinStayLoseShift(),
    axl.APavlov2011(),
    axl.ContriteTitForTat(),
    axl.EvolvedANNNoise05(),
    axl.RiskyQLearner(),
    axl.Random(),
    axl.DBS(),
]

DBS_vs_results = {} 
for noise in [0, 0.01, 0.05, 0.1]:
    tournament = axl.Tournament(
        players=DBS_vs,
        turns=400,
        repetitions=3,
        noise=noise,
        seed=42,
    )
    r = tournament.play(processes=os.cpu_count() - 10)
    DBS_vs_results[noise] = r


for noise, result in DBS_vs_results.items():
    print(f"Noise: {noise}")
    print(result.summarise())
    print("\n")
```

```python
for noise, result in DBS_vs_results.items():
    print(f"Noise: {noise}")
    print(result.ranked_names)
```
