```python
import numpy as np
from pymdp import utils
from pymdp.agent import Agent

# Define dimensions
num_states = [2, 2]  # [My_Last_Action, Their_Last_Action]
num_obs = [2, 2]     # [My_Action_Observation, Their_Action_Observation]
num_controls = [2]   # My possible actions: [Cooperate, Defect]

# Create A matrix (observation model)
A = utils.obj_array_zeros([[num_obs[0], num_states[0], num_states[1]],
                          [num_obs[1], num_states[0], num_states[1]]])

# Create B matrix (transition model)
B = utils.obj_array_zeros([[num_states[0], num_states[0], num_controls[0]],
                          [num_states[1], num_states[1], num_controls[0]]])

# Fill the A matrix - we know our own actions perfectly
A[0][:, :, :] = np.eye(2)[:, np.newaxis, :]

# Initialize the opponent model with uncertainty (uniform prior)
# Their action depends on both our last action and their last action
A[1][:, :, :] = 1/2  # Uniform uncertainty initially

# Set up preferences (C matrix) - prefer mutual cooperation
C = utils.obj_array_zeros(num_obs)
C[1][0] = 3.0   # High preference for observing cooperation
C[1][1] = -1.0  # Low preference for observing defection

# Create agent with learning enabled
agent = Agent(
    A=A, 
    B=B, 
    C=C, 
    policy_len=3,
    inference_algo="VANILLA",  # Standard inference
    use_BMA=True,              # Use Bayesian model averaging
)

# Enable learning of the A matrix (observation model)
agent.learn_A = True
```
