```python
import random
import wandb
```

```python
# Start a new wandb run to track this script.
run = wandb.init(
    # Set the wandb entity where your project will be logged (generally your team name).
    entity="kivalm-University of KwaZulu-Natal",
    # Set the wandb project where this run will be logged.
    project="AIF-NIPD",
    # Track hyperparameters and run metadata.
    config={
        "noise": 0.02,
        "algorithm": "TitForTat",
        "turns": 1000,
    },
)
```

```python
import axelrod as axl

strategies = (axl.TitForTat(), axl.TitForTat())

tournament = axl.Tournament(strategies, turns=1000, noise=0.02)
results = tournament.play(processes=16)
```

```python
import pickle

# dump the results to a file
results_pkl = pickle.dumps(results.json)

# log the results
wandb.log({"results": results_pkl})
```
