```python
import torch
import axelrod as axl
from agents.brl import BayesianQLearner, AggressiveBayesianQLearner, CooperativeBayesianQLearner, CautiousBayesianQLearner, ErraticBayesianQLearner

strategies = [
    axl.Cooperator(),
    axl.Defector(),
    axl.TitForTat(),
    axl.WinStayLoseShift(),
    axl.DBS(),
    axl.ZDExtortion(),
    axl.ArrogantQLearner(),
    axl.RiskyQLearner(),
    axl.APavlov2011(),
    axl.OmegaTFT(),
    axl.Random(),
    BayesianQLearner(),
    AggressiveBayesianQLearner(),
    CooperativeBayesianQLearner(),
    CautiousBayesianQLearner(),
    ErraticBayesianQLearner(),
]
```

```python
tournament = axl.Tournament(strategies, turns=1000, noise=0.00)
results = tournament.play(processes=16)
```

```python
results.summarise()
```
