```python
import sys
from pathlib import Path
import pandas as pd
from experiments import result_utils

# Configuration
dir_agent_map = {
    # "bqlearning": "Jax Bayesian QLearner",
    "qlearning": "Jax QLearner",
    # "actinf": "JAX_AIF",
    # "dbs": "DBS"
    "epsilonaif": "EpsAIF"
}

results_dir = Path("../../results/hyperparameters")

# Analyze all agents across all noise levels
print("Analyzing hyperparameter results...")
results_by_noise = result_utils.analyze_best_agents_per_noise(results_dir, dir_agent_map)

# Get top 10 agents for each noise level
top_agents = result_utils.get_top_n_agents(results_by_noise, n=100)

# Format for display
formatted_results = result_utils.format_results_for_display(top_agents, decimals=2)

print(f"Analysis complete! Found {len(formatted_results)} noise levels.")
```

```python
# Display results as 6 separate tables (one for each noise level)
for noise_level in sorted(formatted_results.keys()):
    print(f"\n{'='*100}")
    print(f"NOISE LEVEL: {noise_level}")
    print(f"{'='*100}\n")
    
    df = formatted_results[noise_level]
    # Add rank column
    df_display = df.copy()
    df_display.insert(0, 'Rank', range(1, len(df_display) + 1))
    
    display(df_display)
```

```python
# Calculate average scores across all noise levels for each agent
print(f"\n{'='*100}")
print("AVERAGE SCORES ACROSS ALL NOISE LEVELS")
print(f"{'='*100}\n")

# Combine all results across noise levels
all_results = []
for noise_level, df in formatted_results.items():
    df_copy = df.copy()
    df_copy['noise_level'] = noise_level
    all_results.append(df_copy)

combined_df = pd.concat(all_results, ignore_index=True)

# Calculate average score for each agent across all noise levels
avg_scores = combined_df.groupby('Hyperparameters')['Mean Score'].mean().sort_values(ascending=False)
avg_scores_df = pd.DataFrame({
    'Hyperparameters': avg_scores.index,
    'Mean Score': avg_scores.values
})

# Add rank column
avg_scores_df.insert(0, 'Rank', range(1, len(avg_scores_df) + 1))

# Reset index for clean display
avg_scores_df = avg_scores_df.reset_index(drop=True)
# avg_scores_df.to_csv('avg_scores_df.csv', index=False)
display(avg_scores_df)
```
